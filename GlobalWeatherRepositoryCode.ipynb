```python
# Delete the impossible values


print('Before cleaning impossible values:', df.shape)
df_clean = df[
(df['temperature_celsius'].between(-89 , 56.7)) &
(df['wind_kph'].between(0 , 408)) &
(df['pressure_mb'].between(870 , 1083.8))
].copy()

print('After cleaning impossible values:', df_clean.shape)

# There were 4 rows with impossible values that we removed.
```

```python
def remove_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return data[(data[column] >= lower) & (data[column] <= upper)]


df_final = df_clean.copy()
for col in ['temperature_celsius', 'wind_kph', 'pressure_mb']:
    df_final = df_final.groupby('country', group_keys = False).apply(
        lambda x: remove_outliers_iqr(x, col)
    ).reset_index(drop = True)

print('After IQR Cleaning:', df_final.shape)
print('Total Removed Rows :', 142483 - df_final.shape[0])
```

```python
def Group_condition(condition):
    
    sunny_keywords = ['sunny' , 'clear' ]
    cloudy_keywords = ['cloudy', 'overcast' , 'mist' , 'fog', 'partly','partly cloudy']
    rainy_keywords = ['rain', 'drizzle' , 'shower']
    stormy_keywords = ['thunder' , 'storm' , 'blizzard']
    snowy_keywords = ['snow' , 'sleet','ice', 'freezing','blowing']
    
    if any(word in condition for word in sunny_keywords):
        return 'Sunny'
    elif any(word in condition for word in stormy_keywords):
        return 'Stormy'
    elif any(word in condition for word in snowy_keywords):
        return 'Snowy'
    elif any(word in condition for word in rainy_keywords):
        return 'Rainy'
    elif any(word in condition for word in cloudy_keywords):
        return 'Cloudy'
    else:
        return 'Other'
    
df_model['condition_group'] = df_model['condition_text'].apply(Group_condition)

print(df_model['condition_group'].value_counts())
print('\nAny Other?',(df_model['condition_group']== 'Other').sum())
```

```python
from sklearn.model_selection import train_test_split

X = df_model.drop(columns=['temperature_celsius',
                            'condition_group',
                            'condition_encoded',
                            'feels_like_celsius'])


y_reg = df_model['temperature_celsius']
y_clf = df_model['condition_encoded']

# Split 1 — Train 80% / Temp 20%
X_train, X_temp, y_reg_train, y_reg_temp, y_clf_train, y_clf_temp = \
    train_test_split(X, y_reg, y_clf,
                    test_size    = 0.2,
                    random_state = 42,
                    stratify     = y_clf)

# Split 2 — Val 10% / Test 10%
X_val, X_test, y_reg_val, y_reg_test, y_clf_val, y_clf_test = \
    train_test_split(X_temp, y_reg_temp, y_clf_temp,
                    test_size    = 0.5,
                    random_state = 42,
                    stratify     = y_clf_temp)

print(f'X_train : {X_train.shape}')
print(f'X_val   : {X_val.shape}')
print(f'X_test  : {X_test.shape}')
print(f'y_reg_train : {y_reg_train.shape}')
print(f'y_reg_val   : {y_reg_val.shape}')
print(f'y_reg_test  : {y_reg_test.shape}')
print(f'y_clf_train : {y_clf_train.shape}')
print(f'y_clf_val   : {y_clf_val.shape}')
print(f'y_clf_test  : {y_clf_test.shape}')
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

# I only did fit_transform on the training set to avoid data leakage, and then I used transform on the validation and test sets to apply the same scaling.

X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

print(f'X_train_scaled shape: {X_train_scaled.shape}')
print(f"X_val_scaled shape: {X_val_scaled.shape}")
print(f'X_test_scaled shape: {X_test_scaled.shape}')
print(f"X_val_scaled   shape: {X_val_scaled.shape}")
print(f'X_test_scaled  shape: {X_test_scaled.shape}')
```

```python
import joblib

# ══════════════════════════════════════════════
# Train Final Models on Train + Validation
# ══════════════════════════════════════════════

# دمج Train و Validation مع بعض
import numpy as np

X_final      = np.concatenate([X_train_scaled, X_val_scaled])
y_reg_final  = np.concatenate([y_reg_train, y_reg_val])
y_clf_final  = np.concatenate([y_clf_train, y_clf_val])

print(f'Final Training Data: {X_final.shape}')

# ══════════════════════════════════════════════
# Train Regression Model
# ══════════════════════════════════════════════
xgb_reg.fit(X_final, y_reg_final)
print('✅ Regression Model Trained!')

# ══════════════════════════════════════════════
# Train Classification Model
# ══════════════════════════════════════════════
xgb_clf.fit(X_final, y_clf_final)
print('✅ Classification Model Trained!')

# ══════════════════════════════════════════════
# Save Models
# ══════════════════════════════════════════════
joblib.dump(xgb_reg,  'weather_regression_model.pkl')
joblib.dump(xgb_clf,  'weather_classification_model.pkl')
joblib.dump(scaler,   'weather_scaler.pkl')
joblib.dump(le,       'weather_label_encoder.pkl')

print('\n✅ All models saved!')
print('   weather_regression_model.pkl')
print('   weather_classification_model.pkl')
print('   weather_scaler.pkl')
print('   weather_label_encoder.pkl')
```

```python
import joblib

# Load Models
xgb_reg  = joblib.load('weather_regression_model.pkl')
xgb_clf  = joblib.load('weather_classification_model.pkl')
scaler   = joblib.load('weather_scaler.pkl')
le       = joblib.load('weather_label_encoder.pkl')

# New data - The same  features 
new_data = pd.DataFrame([{
    'latitude'                    : 30.06,
    'longitude'                   : 31.24,
    'wind_kph'                    : 15.0,
    'wind_degree'                 : 180,
    'pressure_mb'                 : 1013.0,
    'precip_mm'                   : 0.0,
    'humidity'                    : 45,
    'cloud'                       : 20,
    'visibility_km'               : 10.0,
    'uv_index'                    : 6.0,
    'gust_kph'                    : 20.0,
    'air_quality_Carbon_Monoxide' : 300.0,
    'air_quality_Ozone'           : 80.0,
    'air_quality_Nitrogen_dioxide': 10.0,
    'air_quality_Sulphur_dioxide' : 5.0,
    'air_quality_PM2.5'           : 15.0,
    'air_quality_PM10'            : 25.0,
    'air_quality_us-epa-index'    : 1,
    'moon_illumination'           : 50,
    'month'                       : 6,
}])

# Scale
new_data_scaled = scaler.transform(new_data)

# Predict
temp      = xgb_reg.predict(new_data_scaled)[0]
condition = le.inverse_transform(
                xgb_clf.predict(new_data_scaled)
            )[0]

print(f'🌡️  Temperature : {temp:.1f}°C')
print(f'🌤️  Condition   : {condition}')
```
